```python
from pathlib import Path
import pandas as pd

INPUT_CSV = Path(
    r"C:\Users\pkmpp\Documents\GitHub\J26-DS-323\experiments\pasan\step by step\data\solar_5min_with_time_features_03 combine_wether_data.csv"
)

OUTPUT_CSV = INPUT_CSV.with_name(
    f"{INPUT_CSV.stem}_filtered{INPUT_CSV.suffix}"
)

# Verified against device_mapping.csv
EXCLUDED_DEVICES = {
    11: "1801150245270078",
    20: "18050B0239120029",
    23: "10050A021914000",
    30: "1805090245220419",
    36: "1031721264210116",
    43: "1801151255110048",
    46: "3801150252110059",
}

df = pd.read_csv(INPUT_CSV, dtype={"inverterSn": "string"})

# Normalize identifiers for matching without changing the saved column.
serials = df["inverterSn"].str.strip()
remove = serials.isin(EXCLUDED_DEVICES.values())

print("Rows removed per device:")
for device, serial in EXCLUDED_DEVICES.items():
    count = int(serials.eq(serial).sum())
    print(f"  Device {device}: {count:,}")

filtered = df.loc[~remove].copy()

assert not filtered["inverterSn"].str.strip().isin(
    EXCLUDED_DEVICES.values()
).any()

filtered.to_csv(OUTPUT_CSV, index=False)

print(f"\nOriginal rows:     {len(df):,}")
print(f"Removed rows:      {int(remove.sum()):,}")
print(f"Remaining rows:    {len(filtered):,}")
print(f"Remaining devices: {filtered['inverterSn'].str.strip().nunique()}")
print(f"Saved to: {OUTPUT_CSV}")
```
