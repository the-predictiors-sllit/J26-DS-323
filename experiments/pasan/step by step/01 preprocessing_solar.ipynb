### add new data to csv

2. in this code for each file append new data

```python

import pandas as pd

info_by_filename = {
    item["file"].casefold(): item
    for item in infolist
}

updated = []
missing_from_infolist = []
errors = []

for csv_path in chartdata_folder.rglob("*.csv"):
    filename = csv_path.name
    info = info_by_filename.get(filename.casefold())

    if info is None:
        missing_from_infolist.append(filename)
        continue

    try:
        # Read CSV while reducing mixed-type inference warnings
        df = pd.read_csv(csv_path, low_memory=False)

        # Build all metadata columns together
        metadata = pd.DataFrame({
            "file": [filename] * len(df),
            "town": [info["town"]] * len(df),
            "district": [info["district"]] * len(df),
            "pv_capacity_kwp": [info["pv_capacity_kwp"]] * len(df),
            "longitude": [info["longitude"]] * len(df),
            "latitude": [info["latitude"]] * len(df),
        }, index=df.index)

        # Replace existing metadata columns instead of duplicating them
        df = df.drop(
            columns=metadata.columns.intersection(df.columns),
            errors="ignore"
        )

        # Combine the original data and metadata
        df = pd.concat([df, metadata], axis=1)

        # Overwrite the original file
        df.to_csv(csv_path, index=False)

        updated.append(filename)

    except Exception as error:
        errors.append((filename, str(error)))

print(f"Updated: {len(updated)}")
print(f"Missing from infolist: {len(missing_from_infolist)}")
print(f"Errors: {len(errors)}")

print("\nUpdated files:")
print(updated)

print("\nFiles missing from infolist:")
print(missing_from_infolist)

print("\nErrors:")
print(errors)
```

## Round up the 5 minute time interverls

4. this will create a new folder named aligned_5min

```python
from pathlib import Path
import pandas as pd

INPUT_FOLDER = Path(r"C:\Users\pkmpp\Documents\GitHub\J26-DS-323\experiments\pasan\step by step\data\.bigdata")
OUTPUT_FOLDER = Path(r"C:\Users\pkmpp\Documents\GitHub\J26-DS-323\experiments\pasan\step by step\data") / "aligned_5min_01 preprocessing_solar"

OUTPUT_FOLDER.mkdir(parents=True, exist_ok=True)

csv_files = sorted(INPUT_FOLDER.glob("*.csv"))

if not csv_files:
    raise FileNotFoundError(f"No CSV files found in {INPUT_FOLDER}")

report = []

for source in csv_files:
    destination = OUTPUT_FOLDER / source.name

    if destination.exists():
        print(f"SKIPPED: {source.name} — output already exists")
        continue

    try:
        # Preserve long inverter identifiers and original field values.
        df = pd.read_csv(
            source,
            dtype="string",
            encoding="utf-8-sig",
        )

        required = {"inverterSn", "dataTimestamp", "time"}
        missing = required - set(df.columns)

        if missing:
            raise ValueError(f"Missing required columns: {sorted(missing)}")

        # Parse the original Unix timestamp without losing milliseconds.
        epoch_ms = pd.to_numeric(df["dataTimestamp"], errors="coerce")

        timestamp_utc = pd.to_datetime(
            epoch_ms,
            unit="ms",
            utc=True,
            errors="coerce",
        )

        timestamp_local = timestamp_utc.dt.tz_convert("Asia/Colombo")

        # `time` is local Sri Lankan time in your sample.
        recorded_local = (
            pd.to_datetime(
                df["time"],
                format="%Y-%m-%d %H:%M:%S",
                errors="coerce",
            )
            .dt.tz_localize("Asia/Colombo")
        )

        df["timestamp_utc"] = timestamp_utc
        df["timestamp_local"] = timestamp_local

        df["timestamp_invalid"] = (
            timestamp_local.isna() | recorded_local.isna()
        )

        # Allow one second because `time` omits milliseconds.
        df["time_difference_seconds"] = (
            timestamp_local - recorded_local
        ).dt.total_seconds()

        df["timestamp_mismatch"] = (
            df["time_difference_seconds"].abs() > 1
        )

        # Round to the nearest five-minute grid point.
        df["timestamp_5min"] = timestamp_local.dt.round("5min")

        df["alignment_shift_seconds"] = (
            df["timestamp_5min"] - timestamp_local
        ).dt.total_seconds()

        # Flag collisions; retain every original row.
        keys = ["inverterSn", "timestamp_5min"]

        valid_key = (
            df["inverterSn"].notna()
            & df["inverterSn"].str.strip().ne("")
            & df["timestamp_5min"].notna()
        )

        df["duplicate_5min"] = False
        df.loc[valid_key, "duplicate_5min"] = (
            df.loc[valid_key].duplicated(keys, keep=False)
        )

        df = df.sort_values(
            ["inverterSn", "timestamp_local"],
            kind="stable",
        )

        df.to_csv(destination, index=False, encoding="utf-8-sig")

        report.append({
            "file": source.name,
            "rows": len(df),
            "invalid_timestamps": int(df["timestamp_invalid"].sum()),
            "timestamp_mismatches": int(df["timestamp_mismatch"].sum()),
            "rows_with_duplicate_5min": int(df["duplicate_5min"].sum()),
        })

        print(f"SAVED: {source.name} — {len(df):,} rows")

    except (ValueError, OSError, pd.errors.ParserError) as error:
        report.append({
            "file": source.name,
            "error": str(error),
        })
        print(f"FAILED: {source.name} — {error}")

if report:
    pd.DataFrame(report).to_csv(
        OUTPUT_FOLDER / "alignment_report.csv",
        index=False,
        encoding="utf-8-sig",
    )

print(f"\nFinished. Output folder: {OUTPUT_FOLDER}")
```

## remove rows that alarmState = 2 or 3

this indicate something happend to the inverter reading or fault. so we do not need that

```python
from pathlib import Path
import pandas as pd

INPUT_FOLDER = Path(r"C:\Users\pkmpp\Documents\GitHub\J26-DS-323\experiments\pasan\step by step\data\aligned_5min_01 preprocessing_solar")
OUTPUT_FOLDER = Path(r"C:\Users\pkmpp\Documents\GitHub\J26-DS-323\experiments\pasan\step by step\data") / "filtered_training(no alarmState 2 or 3)_01 preprocessing_solar"

if not INPUT_FOLDER.is_dir():
    raise FileNotFoundError(f"Folder not found: {INPUT_FOLDER}")

OUTPUT_FOLDER.mkdir(exist_ok=True)

for source in sorted(INPUT_FOLDER.glob("*.csv")):
    # Preserve identifiers and original values.
    df = pd.read_csv(source, dtype="string", encoding="utf-8-sig")

    if "alarmState" not in df.columns:
        print(f"SKIPPED: {source.name} — no alarmState column")
        continue

    status = pd.to_numeric(df["alarmState"], errors="coerce")

    # Remove statuses 2 and 3; retain the alarmState column.
    filtered = df.loc[~status.isin([2, 3])].copy()

    filtered.to_csv(
        OUTPUT_FOLDER / source.name,
        index=False,
        encoding="utf-8-sig",
    )

    print(
        f"{source.name}: "
        f"removed {len(df) - len(filtered):,}, "
        f"kept {len(filtered):,}"
    )

print(f"\nSaved to: {OUTPUT_FOLDER}")
```

## remove colums timestamp_invalid, timestamp_mismatch, or duplicate_5min is True

```python
from pathlib import Path
import pandas as pd

INPUT_FOLDER = Path(r"C:\Users\pkmpp\Documents\GitHub\J26-DS-323\experiments\pasan\step by step\data\filtered_training(no alarmState 2 or 3)_01 preprocessing_solar")
OUTPUT_FOLDER = Path(r"C:\Users\pkmpp\Documents\GitHub\J26-DS-323\experiments\pasan\step by step\data") / "cleaned1_01 preprocessing_solar"

FLAG_COLUMNS = [
    "timestamp_invalid",
    "timestamp_mismatch",
    "duplicate_5min",
]

DROP_COLUMNS = [
    "timeStr",
    "time",
    "eToday",
    "currentState",
    "warningInfoData",
    "alarmState",
    "pvAndAcCoupledPower",
    "productModel",
    "alignment_shift_seconds",
    "time_difference_seconds",
    *FLAG_COLUMNS,
]

if not INPUT_FOLDER.is_dir():
    raise FileNotFoundError(f"Folder not found: {INPUT_FOLDER}")

OUTPUT_FOLDER.mkdir(exist_ok=True)

for source in sorted(INPUT_FOLDER.glob("*.csv")):
    df = pd.read_csv(source, dtype="string", encoding="utf-8-sig")

    missing_flags = [c for c in FLAG_COLUMNS if c not in df.columns]
    if missing_flags:
        print(f"SKIPPED: {source.name} — missing flags: {missing_flags}")
        continue

    # Accept True/False and 1/0 representations.
    flags = df[FLAG_COLUMNS].apply(
        lambda col: col.str.strip().str.lower()
    )

    # Avoid dropping the flags before unresolved values are checked.
    if not flags.isin(["true", "false", "1", "0"]).all().all():
        print(f"SKIPPED: {source.name} — missing or unrecognized flag values")
        continue

    remove_rows = flags.isin(["true", "1"]).any(axis=1)

    cleaned = (
        df.loc[~remove_rows]
        .drop(columns=DROP_COLUMNS, errors="ignore")
    )

    cleaned.to_csv(
        OUTPUT_FOLDER / source.name,
        index=False,
        encoding="utf-8-sig",
    )

    print(
        f"{source.name}: removed {remove_rows.sum():,} rows; "
        f"kept {len(cleaned):,} rows and {len(cleaned.columns)} columns"
    )

print(f"\nSaved to: {OUTPUT_FOLDER}")
```

## remove all other rows that does not belong to 06:00:00+05:30 - 19:00:00+05:30 range

```python
from pathlib import Path
import pandas as pd

INPUT_FOLDER = Path(r"C:\Users\pkmpp\Documents\GitHub\J26-DS-323\experiments\pasan\step by step\data\cleaned1_01 preprocessing_solar")
OUTPUT_FOLDER = Path(r"C:\Users\pkmpp\Documents\GitHub\J26-DS-323\experiments\pasan\step by step\data") / "daytime_0600_1900_01 preprocessing_solar"

if not INPUT_FOLDER.is_dir():
    raise FileNotFoundError(f"Folder not found: {INPUT_FOLDER}")

OUTPUT_FOLDER.mkdir(exist_ok=True)

for source in sorted(INPUT_FOLDER.glob("*.csv")):
    df = pd.read_csv(source, dtype="string", encoding="utf-8-sig")

    if "timestamp_5min" not in df.columns:
        print(f"SKIPPED: {source.name} — missing timestamp_5min")
        continue

    # Aligned timestamps already contain their timezone offset.
    local_time = pd.to_datetime(
        df["timestamp_5min"],
        utc=True,
        errors="coerce",
    ).dt.tz_convert("Asia/Colombo")

    seconds = (
        local_time.dt.hour * 3600
        + local_time.dt.minute * 60
        + local_time.dt.second
    )

    keep = seconds.between(6 * 3600, 19 * 3600, inclusive="both")
    filtered = df.loc[keep].copy()

    filtered.to_csv(
        OUTPUT_FOLDER / source.name,
        index=False,
        encoding="utf-8-sig",
    )

    print(
        f"{source.name}: kept {len(filtered):,}; "
        f"removed {len(df) - len(filtered):,}; "
        f"invalid timestamps {local_time.isna().sum():,}"
    )

print(f"\nSaved to: {OUTPUT_FOLDER}")
```

## combine data into one file

```python
from pathlib import Path
import pandas as pd

INPUT_FOLDER = Path(r"C:\Users\pkmpp\Documents\GitHub\J26-DS-323\experiments\pasan\step by step\data\daytime_0600_1900_01 preprocessing_solar")
OUTPUT_FILE = Path(r"C:\Users\pkmpp\Documents\GitHub\J26-DS-323\experiments\pasan\step by step\data") / "final_solaronly_combined_01 preprocessing_solar.csv"

# Exclude the output so rerunning cannot combine it with itself.
files = sorted(
    path for path in INPUT_FOLDER.glob("*.csv")
    if path.name.lower() != OUTPUT_FILE.name.lower()
)

if not files:
    raise FileNotFoundError(f"No input CSV files found in {INPUT_FOLDER}")

# Check that every file has the same columns before writing.
columns = pd.read_csv(
    files[0], nrows=0, encoding="utf-8-sig"
).columns.tolist()

for path in files:
    header = pd.read_csv(
        path, nrows=0, encoding="utf-8-sig"
    ).columns.tolist()

    if set(header) != set(columns):
        raise ValueError(f"Columns do not match: {path.name}")

temporary = OUTPUT_FILE.with_suffix(".csv.tmp")
total_rows = 0

try:
    with temporary.open("w", encoding="utf-8-sig", newline="") as output:
        # Write the header once, including when inputs have no rows.
        pd.DataFrame(columns=columns).to_csv(output, index=False)

        for path in files:
            file_rows = 0

            with pd.read_csv(
                path,
                dtype="string",
                encoding="utf-8-sig",
                chunksize=50_000,
            ) as reader:
                for chunk in reader:
                    chunk[columns].to_csv(
                        output, index=False, header=False
                    )
                    file_rows += len(chunk)

            total_rows += file_rows
            print(f"Added: {path.name} — {file_rows:,} rows")

    temporary.replace(OUTPUT_FILE)

finally:
    if temporary.exists():
        temporary.unlink()

print(f"\nCombined {len(files)} files: {total_rows:,} rows")
print(f"Saved to: {OUTPUT_FILE}")
```
