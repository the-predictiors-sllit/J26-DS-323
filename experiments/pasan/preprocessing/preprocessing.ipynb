```python
from pathlib import Path


chartdata_folder =  Path(".chartdata/")

existing_files = {
    path.name.casefold(): path 
    for path in chartdata_folder.rglob("*")
    if path.is_file()
}

found = []
missing = []

for item in infolist:
    filename = item["file"]
    matching_path = existing_files.get(filename.casefold())

    if matching_path:
        found.append((filename, matching_path))
    else:
        missing.append(filename)

print(f"Found: {len(found)}")
print(f"Missing: {len(missing)}")
```

### add new data to csv

```python
info_by_filename = {
    item["file"].casefold(): item
    for item in infolist
}



updated = []
missing_from_infolist = []
errors = []

for csv_path in chartdata_folder.rglob("*.csv"):
    filename = csv_path.name
    info = info_by_filename.get(filename.casefold())

    if info is None:
        missing_from_infolist.append(filename)
        continue

    try:
        df = pd.read_csv(csv_path)

        # Add the filename and metadata to every row
        df["file"] = filename
        df["town"] = info["town"]
        df["district"] = info["district"]
        df["pv_capacity_kwp"] = info["pv_capacity_kwp"]

        # Overwrite the original file
        df.to_csv(csv_path, index=False)

        updated.append(filename)

    except Exception as error:
        errors.append((filename, str(error)))


print(updated)
print(missing_from_infolist)
print(errors)
```
