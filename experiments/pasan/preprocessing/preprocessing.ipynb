```python
from pathlib import Path


chartdata_folder =  Path(".chartdata/")

existing_files = {
    path.name.casefold(): path 
    for path in chartdata_folder.rglob("*")
    if path.is_file()
}

found = []
missing = []

for item in infolist:
    filename = item["file"]
    matching_path = existing_files.get(filename.casefold())

    if matching_path:
        found.append((filename, matching_path))
    else:
        missing.append(filename)

print(f"Found: {len(found)}")
print(f"Missing: {len(missing)}")
```

### add new data to csv

```python
import pandas as pd


info_by_filename = {
    item["file"].casefold(): item
    for item in infolist
}



updated = []
missing_from_infolist = []
errors = []

for csv_path in chartdata_folder.rglob("*.csv"):
    filename = csv_path.name
    info = info_by_filename.get(filename.casefold())

    if info is None:
        missing_from_infolist.append(filename)
        continue

    try:
        df = pd.read_csv(csv_path)

        # Add the filename and metadata to every row
        df["file"] = filename
        df["town"] = info["town"]
        df["district"] = info["district"]
        df["pv_capacity_kwp"] = info["pv_capacity_kwp"]
        df["longitude"] = info["longitude"]
        df["latitude"] = info["latitude"]

        # Overwrite the original file
        df.to_csv(csv_path, index=False)

        updated.append(filename)

    except Exception as error:
        errors.append((filename, str(error)))


print(updated)
print(missing_from_infolist)
print(errors)
```

### Combine all data into one csv

```python
from pathlib import Path
import pandas as pd

INPUT_FOLDER = Path(r"C:\Users\pkmpp\Documents\GitHub\J26-DS-323\experiments\pasan\preprocessing\.chartdata")
OUTPUT_FILE = INPUT_FOLDER /"combined.csv"

OUTPUT_COLUMNS = [
    "device_number",
    "file",
    "query_date",
    "timeStr",
    "power",
    "powerUnit",
    "pvAndAcCoupledPowerList",
    "pv_capacity_kwp",
    "timestamp",
    "town",
    "district",
    "longitude",
    "latitude",
]

all_data = []

csv_files = sorted(INPUT_FOLDER.glob("*.csv"))


for device_number, csv_file in enumerate(csv_files, start=1):
    df = pd.read_csv(csv_file)

    # Add device number starting from 1 for each CSV file
    df["device_number"] = device_number

    # Store the original file name
    df["file"] = csv_file.name

    # Add missing columns as empty values
    for column in OUTPUT_COLUMNS:
        if column not in df.columns:
            df[column] = pd.NA

    # Keep only the required columns in the required order
    df = df[OUTPUT_COLUMNS]

    all_data.append(df)

if all_data:
    combined_df = pd.concat(all_data, ignore_index=True)
    combined_df.to_csv(OUTPUT_FILE, index=False)
    print(f"Combined {len(csv_files)} CSV files into:")
    print(OUTPUT_FILE)
else:
    print("No CSV files found.")
```
